# 2. Demographics Statistics

```r
source("../util/packages.R")
source("../util/variables.R")
ALL_PARTICIPANTS_METADATA <- read_rds("../R-table/ALL_PARTICIPANTS_METADATA.rds")
```

```r
summary_tbl <- ALL_PARTICIPANTS_METADATA |>
  distinct(pseudonym, .keep_all = TRUE) |>
  summarise(
    n_participants = n_distinct(pseudonym),
    n_male = sum(biological_sex == "Male"),
    n_female = sum(biological_sex == "Female"),
    bmi_mean = mean(as.numeric(bmi), na.rm = TRUE),
    bmi_sd = sd(as.numeric(bmi), na.rm = TRUE)
  )

summary_tbl

# Age group counts
age_counts <- ALL_PARTICIPANTS_METADATA |>
  distinct(pseudonym, .keep_all = TRUE) |>
  count(age_group)

age_counts
```
